#How This Works
##Function Definitions:
* Each function uses Python’s re module (with re.IGNORECASE for case-insensitive matching) to search for specific keywords or phrases.
* If a pattern is matched, the corresponding label is added to a list; multiple matches are concatenated with ", ". If no matches are found, a default value (like "Unclassified" or "No") is returned.

##File Upload & DataFrame Creation:
The code uses google.colab.files.upload() to let you upload a CSV file. It then reads the CSV into a Pandas DataFrame.

##Applying Functions:
Each new field is computed by applying the corresponding function to the “Keyword” column, and the results are stored in new columns.

##Result:
The updated DataFrame is displayed (using df.head()) so you can verify the new fields.

**You can customize any of the regex patterns or default values to better suit your needs.**

```python
import pandas as pd
import re
from google.colab import files

# Define functions for each field

def query_length(query):
    """Calculates the number of words in the query."""
    if not isinstance(query, str):
        return 0
    return len(query.strip().split())

def main_intent_category(query):
    """Determines the main intent category (Informational, Transactional, Commercial, Navigational)."""
    categories = []
    # Informational
    informational_pattern = r"\b(?:how|why|does|which|when|who|guide|tutorial|learn|examples|resource|ideas|tips|difference between|strategies|ways|example|directory|news|basics|concepts|troubleshoot|resolve|risks|beginners|upgrade|improve|optimize|case study|best practices|templates|pros and cons|analysis|articles|audit|benefits|blogs|build|calculate|campaign|choose|common questions|consider|cost|create|definition|definitive guide|earn|easy steps|essential|events|explainer video|factors|find|fix|grow|hacks|how to|ideal|importance of|increase|infographic|information|meaning|measure|metrics|mistakes|myths|opportunities|popular|practices|questions|remove|set up|shortcuts|stats|strategy|study|successful|tactics|techniques|terms|things to know|time|trends|tricks|types|updates|well explained|tutorials|framework|checklist|process|plan|test|glossary|sample|survey|exercises|report|certification|webinar|calendar|quiz|directories|to do list|design basics)\b"
    if re.search(informational_pattern, query, re.IGNORECASE):
         categories.append("Informational")
    # Transactional
    transactional_pattern = r"\b(?:buy|purchase|order|price|quote|for sale|sale|deal|offers|discount|coupon|promo code|voucher|bargain|cost|pricing|cheap|affordable|expensive|premium|near me|online|e-commerce|shopping cart|checkout|shop|store|retailer|dealer|supplier|marketplace|auction|bid|buy now|pay|payment|finance|leasing|rent|rental|subscribe|subscription|free trial|book|reserve|enquire|inquiry|quote|estimate|cost calculator|shipping|delivery|available|in stock|out of stock|pre-order|backorder|special offer|limited time offer|flash sale|clearance|wholesale|bulk buy|group buy|membership|reward|loyalty|gift card|cashback|trade-in|upgrade|renew|checkout|secure checkout)\b"
    if re.search(transactional_pattern, query, re.IGNORECASE):
         categories.append("Transactional")
    # Commercial
    commercial_pattern = r"\b(?:best|top|vs|review|compare|comparison|ratings|recommendation|buyer's guide|alternative|feature|benefit|cost-effective|affordable|premium|quality|deal|offers|savings|discount|value|choice|selection|rankings|user ratings|testimonials|feedback|pros|cons|evaluation|assessment|analysis|benchmark|brand|models|versions|upgrade|update|launch|new|latest|innovative|technology|solutions|providers|vendors|suppliers|applications|software|platform|systems|tools|equipment|machinery|accessories|devices|gadgets|electronics|appliances|products|materials|suppli|kit|package|bundle)\b"
    if re.search(commercial_pattern, query, re.IGNORECASE):
         categories.append("Commercial")
    # Navigational
    navigational_pattern = r"\b(?:login|register|homepage|website|account|portal|open|access|find|locate|contact us|help desk|FAQ|support|customer service|map|location|directions|hours|site|navigate|URL|service center|branch|office|near me|address)\b"
    if re.search(navigational_pattern, query, re.IGNORECASE):
         categories.append("Navigational")
    return ", ".join(sorted(set(categories))) if categories else "Unclassified"

def query_type(query):
    """Determines the query type (Do, Know, or Go)."""
    types = []
    do_pattern = r"\b(?:buy|purchase|order|download|subscribe|install|upgrade|reserve|book|sign up|apply)\b"
    if re.search(do_pattern, query, re.IGNORECASE):
         types.append("Do")
    know_pattern = r"\b(?:what|why|when|who|which|how(?: to)?|information|meaning|definition|difference|explained|example|types|benefits|risks|importance|case study|analysis|strategy|guide|explainer|tutorial|tips)\b"
    if re.search(know_pattern, query, re.IGNORECASE):
         types.append("Know")
    go_pattern = r"\b(?:find|locate|map|directions|near me|address|navigate|where|office|branch|service center|homepage|site|portal|visit|store|contact)\b"
    if re.search(go_pattern, query, re.IGNORECASE):
         types.append("Go")
    return ", ".join(sorted(set(types))) if types else "Unclassified"

def geo_location_relevant(query):
    """Checks if the query is geo-location relevant."""
    pattern = r"\b(?:near me|location|map|directions|address|store|branch|office|service center|city|state|country|region|zip code|postal code|local|nearby)\b"
    return "Yes" if re.search(pattern, query, re.IGNORECASE) else "No"

def explicit_desired_content_format(query):
    """Specifies if the user seeks a particular content format."""
    formats = []
    video_pattern = r"\b(?:video|tutorial|explainer|short video|tiktok|youtube)\b"
    if re.search(video_pattern, query, re.IGNORECASE):
         formats.append("Yes - video")
    infographic_pattern = r"\b(?:infographic|chart|graph|diagram)\b"
    if re.search(infographic_pattern, query, re.IGNORECASE):
         formats.append("Yes - infographic")
    template_pattern = r"\b(?:template|sample|checklist|list)\b"
    if re.search(template_pattern, query, re.IGNORECASE):
         formats.append("Yes - template")
    document_pattern = r"\b(?:case study|report|whitepaper|pdf|brochure|catalog|study)\b"
    if re.search(document_pattern, query, re.IGNORECASE):
         formats.append("Yes - document")
    webinar_pattern = r"\b(?:webinar|presentation|seminar)\b"
    if re.search(webinar_pattern, query, re.IGNORECASE):
         formats.append("Yes - webinar")
    podcast_pattern = r"\b(?:podcast|audio|radio)\b"
    if re.search(podcast_pattern, query, re.IGNORECASE):
         formats.append("Yes - podcast")
    ebook_pattern = r"\b(?:ebook|book|manual)\b"
    if re.search(ebook_pattern, query, re.IGNORECASE):
         formats.append("Yes - eBook")
    return ", ".join(sorted(set(formats))) if formats else "No"

def explicit_knowledge_level(query):
    """Determines the explicit knowledge level (Beginner, Intermediate, Expert)."""
    levels = []
    beginner_pattern = r"\b(?:beginner|basic|intro|starter|simple|easy|fundamental|101|first time)\b"
    if re.search(beginner_pattern, query, re.IGNORECASE):
         levels.append("Beginner")
    intermediate_pattern = r"\b(?:intermediate|moderate|average|standard|middle level)\b"
    if re.search(intermediate_pattern, query, re.IGNORECASE):
         levels.append("Intermediate")
    expert_pattern = r"\b(?:advanced|expert|pro|professional|detailed|in[- ]depth)\b"
    if re.search(expert_pattern, query, re.IGNORECASE):
         levels.append("Expert")
    return ", ".join(sorted(set(levels))) if levels else "Unclassified"

def explicit_user_persona(query):
    """Identifies the explicit user persona (e.g., Student/Educator, Professional, Casual User, etc.)."""
    personas = []
    student_pattern = r"\b(?:student|learner|teacher|educator|researcher|professor|trainer|academic|course|scholar|class|college|university|tutor|homework|study|assignment)\b"
    if re.search(student_pattern, query, re.IGNORECASE):
         personas.append("Student/Educator")
    professional_pattern = r"\b(?:professional|manager|consultant|business|enterprise|corporate|industry|employee|freelancer|executive|director|entrepreneur|career|office|workplace|job|resume|cv)\b"
    if re.search(professional_pattern, query, re.IGNORECASE):
         personas.append("Professional")
    casual_pattern = r"\b(?:casual|hobbyist|enthusiast|individual|personal|diy|home|family|friend|weekend|daily|ordinary|simple)\b"
    if re.search(casual_pattern, query, re.IGNORECASE):
         personas.append("Casual User")
    affluent_pattern = r"\b(?:affluent|wealthy|rich|moneyed|millionaire|billionaire|upper class|elite|well[ -]?off|high net worth|investor)\b"
    if re.search(affluent_pattern, query, re.IGNORECASE):
         personas.append("Affluent")
    cheapskate_pattern = r"\b(?:cheapskate|frugal|stingy|budget[ -]?conscious|cheap|inexpensive|discount|deal|bargain|economical|low cost|affordable|coupon|sale|clearance)\b"
    if re.search(cheapskate_pattern, query, re.IGNORECASE):
         personas.append("Cheapskate")
    luxury_pattern = r"\b(?:luxury|premium|high[ -]?end|exclusive|designer|upscale|deluxe|opulent|lavish|extravagant|bespoke|first[ -]?class|elegant|fancy)\b"
    if re.search(luxury_pattern, query, re.IGNORECASE):
         personas.append("Luxury")
    innovator_pattern = r"\b(?:innovator|innovative|disruptor|cutting[ -]?edge|pioneering|trailblazer|visionary|tech[ -]?savvy|groundbreaking|inventor|avant[ -]?garde|revolutionary|futuristic|game changer)\b"
    if re.search(innovator_pattern, query, re.IGNORECASE):
         personas.append("Innovator")
    return ", ".join(sorted(set(personas))) if personas else "Unclassified"

def device_relevant(query):
    """Checks if the query is device-relevant."""
    pattern = r"\b(?:mobile|phone|cellphone|tablet|app|android|ios|iphone|smartphone|samsung|apple|pixel|nokia|oneplus|huawei|desktop|pc|laptop|mac|windows|device|gadget|computer|notebook|ultrabook|wearable|smartwatch|ipad|lg|windows phone)\b"
    return "Yes" if re.search(pattern, query, re.IGNORECASE) else "No"

def implicit_intent_problem(query):
    """Indicates if there is an implicit problem or struggle in the query."""
    pattern = r"\b(?:how to fix|how can I fix|how do I fix|fix(?:ing)?|resolve|troubleshoot|solve|problem(?:s)?|issue(?:s)?|error(?:s)?|error code(?:s)?|failure(?:s)?|crash(?:es)?|repair(?:ing)?|why not|can't|cannot|won't|doesn't|not work|unresponsive|stuck|difficult(?:y)?|struggle(?:s)?|trouble(?:s)?|help|assist(?:ance)?|support|advice|what(?:'s| is) wrong|malfunction(?:ing)?|broken|bug(?:s)?|glitch(?:es)?|defective|unable|freeze(?:s|n)?|hang(?:s|ed)?|lag(?:ging)?|overheat(?:ing)?|no response|no power|no signal|can't load|can't update|can't connect)\b"
    return "Yes" if re.search(pattern, query, re.IGNORECASE) else "No"

def context(query):
    """Determines additional context regarding timing/urgency and mood/emotion."""
    time_pattern = r"\b(?:urgent|now|today|quick|immediately|asap|recent|current|hurry|rush|emergency|priority|fast|rapid|soon)\b"
    mood_pattern = r"\b(?:mood|excited|frustrated|stressed|sad|happy|angry|mad|furious|irate|annoyed|upset|disappointed|depressed|anxious|overwhelmed|exasperated|infuriated|f\*\*k|fuck)\b"
    time_match = re.search(time_pattern, query, re.IGNORECASE)
    mood_match = re.search(mood_pattern, query, re.IGNORECASE)
    if time_match and mood_match:
         return "Yes - Time (Urgency) & Mood (Emotion)"
    elif time_match:
         return "Yes - Time (Urgency)"
    elif mood_match:
         return "Yes - Mood (Emotion)"
    else:
         return "No"

# Upload the CSV file from your computer
uploaded = files.upload()

# Adjust the filename to your uploaded file's name
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# Assume the CSV has a column named "Keyword"
df["Query Length"] = df["Keyword"].apply(query_length)
df["Main Intent Category"] = df["Keyword"].apply(main_intent_category)
df["Query Type"] = df["Keyword"].apply(query_type)
df["Geo-Location Relevant?"] = df["Keyword"].apply(geo_location_relevant)
df["Explicit Desired Content Format"] = df["Keyword"].apply(explicit_desired_content_format)
df["Explicit Knowledge Level"] = df["Keyword"].apply(explicit_knowledge_level)
df["Explicit User Persona"] = df["Keyword"].apply(explicit_user_persona)
df["Device Relevant?"] = df["Keyword"].apply(device_relevant)
df["Implicit Intent - Problem/Struggle"] = df["Keyword"].apply(implicit_intent_problem)
df["Context"] = df["Keyword"].apply(context)

# Display the first few rows of the updated DataFrame
df.head()
```

```python
# Save the updated DataFrame to a CSV file and download it
df.to_csv('updated_file.csv', index=False)
from google.colab import files
files.download('updated_file.csv')
```
