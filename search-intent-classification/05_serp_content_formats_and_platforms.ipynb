#Identify user-desired content formats from SERP features


```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

def upload_and_analyze_serp_features():
    # upload your master file
    uploaded = files.upload()

    # Load the uploaded file into a pandas DataFrame
    file_name = list(uploaded.keys())[0]
    data = pd.read_csv(file_name)

    # Checks if 'SERP features' column exists
    if 'SERP features' not in data.columns:
        print("The uploaded file does not have a 'SERP features' column.")
        return

    # Process the SERP features column
    serp_features_series = data['SERP features'].dropna()  # Remove NaN values
    serp_features_list = serp_features_series.str.split(", ")  # Split strings into lists
    all_serp_features = [feature for sublist in serp_features_list for feature in sublist]  # Flatten the list

    # Count occurrences of each unique SERP feature
    serp_feature_counts = pd.Series(all_serp_features).value_counts()

    # Create a DataFrame for display
    serp_feature_table = pd.DataFrame({
        'SERP Feature': serp_feature_counts.index,
        'Count': serp_feature_counts.values
    })

    # Display the table
    print("SERP Feature Counts:")
    print(serp_feature_table)

    # Visualization: Bar chart for SERP features
    plt.figure(figsize=(12, 6))
    serp_feature_counts.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title("Frequency of SERP Features")
    plt.xlabel("SERP Features")
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# Call the function to execute in Google Colab
upload_and_analyze_serp_features()
```

# Turn your SERP Dictionary CSV into a Python Dictionary

```python
from google.colab import files
import pandas as pd

def csv_to_serp_dictionary_colab():
    """
    Allows the user to upload a CSV file in Google Colab and converts it into a Python dictionary for SERP snippet mappings.

    The CSV should have two columns:
        1. 'SERP snippet' - The name of the SERP feature.
        2. 'Description' - The desired user content format or description for the SERP feature.

    Returns:
        dict: A dictionary where keys are SERP snippets and values are their descriptions.
    """
    # Prompt the user to upload a file
    print("Please upload your CSV file.")
    uploaded = files.upload()

    # Process the uploaded file
    for filename in uploaded.keys():
        data = pd.read_csv(filename)
        break  # Process only the first uploaded file

    # Check if the necessary columns exist
    if 'SERP snippet' not in data.columns or 'Description' not in data.columns:
        raise ValueError("The uploaded CSV must contain 'SERP snippet' and 'Description' columns.")

    # Convert the DataFrame into a dictionary
    serp_dict = dict(zip(data['SERP snippet'], data['Description']))

    return serp_dict

# Example usage
serp_dictionary = csv_to_serp_dictionary_colab()

# Display the resulting dictionary
print(serp_dictionary)
len(serp_dictionary)
```

# Mark up each query with the desired content formats based on the SERP Features


```python
import pandas as pd
from google.colab import files

def process_serp_features_file(serp_snippet_to_content_type):
    """
    Processes an uploaded file, maps SERP features to desired content formats,
    and saves the updated file for download.

    Args:
        serp_snippet_to_content_type (dict): A dictionary mapping SERP snippets
        to their corresponding desired content formats.
    """
    # Allow user to upload a file
    print("Please upload your CSV file containing SERP features.")
    uploaded = files.upload()

    # Load the uploaded file into a pandas DataFrame
    for filename in uploaded.keys():
        data = pd.read_csv(filename)
        break  # Only process the first uploaded file

    # Normalize the dictionary keys for matching
    serp_snippet_to_content_type_normalized = {k.strip().lower(): v for k, v in serp_snippet_to_content_type.items()}

    # Define a function to map SERP features to desired content formats
    def map_serp_features_to_content_formats(serp_features):
        if pd.isna(serp_features):
            return ""
        features = [feature.strip().lower() for feature in serp_features.split(", ")]  # Normalize input
        content_formats = [
            serp_snippet_to_content_type_normalized.get(feature, "Unknown content format.")
            for feature in features
        ]
        return "\n".join(content_formats)

    # Check if the 'SERP features' column exists
    if 'SERP features' not in data.columns:
        raise ValueError("The uploaded file does not contain a 'SERP features' column.")

    # Apply the mapping function to the 'SERP features' column
    data['Desired Content Formats'] = data['SERP features'].apply(map_serp_features_to_content_formats)

    # Save the updated DataFrame to a CSV file
    output_file = "expanded_serp_features.csv"
    data.to_csv(output_file, index=False)

    # Automatically trigger download of the file
    files.download(output_file)

process_serp_features_file(serp_dictionary)
```

## Turn your SERP Dictionary  CSV into a Python Dictionary for Content platforms

```python
from google.colab import files
import pandas as pd

def csv_to_serp_platform_dictionary_colab():
    """
    Allows the user to upload a CSV file in Google Colab and converts it into a Python dictionary for SERP snippet mappings.

    The CSV should have two columns:
        1. 'SERP snippet' - The name of the SERP feature.
        2. 'Content platform' - The most likely content platform served for the SERP feature.

    Returns:
        dict: A dictionary where keys are SERP snippets and values are their likely content platforms.
    """
    # Prompt the user to upload a file
    print("Please upload your CSV file.")
    uploaded = files.upload()

    # Process the uploaded file
    for filename in uploaded.keys():
        data = pd.read_csv(filename)
        break  # Process only the first uploaded file

    # Check if the necessary columns exist
    if 'SERP snippet' not in data.columns or 'Content platform' not in data.columns:
        raise ValueError("The uploaded CSV must contain 'SERP snippet' and 'Content platform' columns.")

    # Convert the DataFrame into a dictionary
    serp_platform_dict = dict(zip(data['SERP snippet'], data['Content platform']))

    return serp_platform_dict

# Example usage
serp_platform_dictionary = csv_to_serp_platform_dictionary_colab()

# Display the resulting dictionary
print(serp_platform_dictionary)
print("Number of entries:", len(serp_platform_dictionary))

```

# Mark up each query with the content platforms Google is most likely to serve based on the SERP Features


```python
import pandas as pd
from google.colab import files

def add_served_content_platforms(serp_platform_dictionary):
    """
    Processes an uploaded file, adds a column for served content platforms,
    and saves the updated file for download.

    Args:
        serp_platform_dictionary (dict): A dictionary mapping SERP snippets
        to their corresponding served content platforms.
    """
    # Allow user to upload a file
    print("Please upload your CSV file containing SERP features.")
    uploaded = files.upload()

    # Load the uploaded file into a pandas DataFrame
    for filename in uploaded.keys():
        data = pd.read_csv(filename)
        break  # Only process the first uploaded file

    # Normalize the platform dictionary for robust matching
    serp_platform_dictionary_normalized = {k.strip().lower(): v for k, v in serp_platform_dictionary.items()}

    # Define a function to map SERP features to served content platforms
    def map_serp_features_to_platforms(serp_features):
        if pd.isna(serp_features):
            return ""
        features = [feature.strip().lower() for feature in serp_features.split(", ")]  # Normalize input
        platforms = [
            serp_platform_dictionary_normalized.get(feature, "Unknown platform.")
            for feature in features
        ]
        return "\n".join(platforms)

    # Apply the mapping function to add the 'Served Content Platforms' column
    if 'SERP features' not in data.columns:
        raise ValueError("The uploaded file does not contain a 'SERP features' column.")

    data['Served Content Platforms'] = data['SERP features'].apply(map_serp_features_to_platforms)

    # Save the updated DataFrame to a CSV file
    output_file = "expanded_serp_features_with_platforms.csv"
    data.to_csv(output_file, index=False)

    # Automatically trigger download of the file
    files.download(output_file)

add_served_content_platforms(serp_platform_dictionary)
```

# Visualize

Note that this function below requires that your output files from the two functions above are still in your local Google colab files. If they are not, either edit the function to take user input instead, add them in the local drive manually, or re-run the previous steps.

```python
import pandas as pd
import matplotlib.pyplot as plt

def process_and_visualize_serp_features_fixed_split():
    """
    Processes two files already in the local Google Colab environment, visualizes data for
    SERP features, desired content formats (with corrected splitting), and served content platforms,
    and outputs three graphs and tables.
    """
    # File paths for the two datasets
    file1_path = 'expanded_serp_features_with_platforms.csv'  # File with Served Content Platforms
    file2_path = 'expanded_serp_features.csv'  # File with Desired Content Formats

    # Load the two files into pandas DataFrames
    data_with_platforms = pd.read_csv(file1_path)
    data_without_platforms = pd.read_csv(file2_path)

    def visualize_and_analyze_columns(data, column_name, title_suffix, split_by=", ", aggregate=True):
        """
        Processes and visualizes the data for a specified column.

        Args:
            data (pd.DataFrame): DataFrame containing the data.
            column_name (str): Name of the column to process and visualize.
            title_suffix (str): Suffix for the plot and table title.
            split_by (str): Delimiter to split entries in the column.
            aggregate (bool): Whether to aggregate the data before visualization.
        """
        if column_name not in data.columns:
            print(f"The column '{column_name}' does not exist in the dataset.")
            return None

        # Process the specified column
        column_series = data[column_name].dropna()  # Remove NaN values
        column_list = column_series.str.split(split_by)  # Split strings into lists
        all_column_entries = [entry.strip() for sublist in column_list for entry in sublist]  # Flatten the list

        # Count occurrences of each unique entry
        column_counts = pd.Series(all_column_entries).value_counts()

        if not aggregate:
            column_counts = column_series.value_counts()  # Use raw series count if not aggregating

        # Create a DataFrame for display
        column_table = pd.DataFrame({
            f'{column_name}': column_counts.index,
            'Count': column_counts.values
        })

        # Visualization: Bar chart for the column entries
        plt.figure(figsize=(12, 6))
        column_counts.head(20).plot(kind='bar', color='skyblue', edgecolor='black')
        plt.title(f"Frequency of {column_name} ({title_suffix})")
        plt.xlabel(column_name)
        plt.ylabel("Count")
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()

        return column_table

    # Analyze and visualize SERP Features
    print("Analyzing: SERP Features Column (From expanded_serp_features_with_platforms.csv)")
    serp_features_table = visualize_and_analyze_columns(data_with_platforms, 'SERP features', 'SERP Features')

    # Analyze and visualize Desired Content Formats (split by "." and newline)
    print("Analyzing: Desired Content Formats Column (From expanded_serp_features.csv)")
    desired_content_formats_table = visualize_and_analyze_columns(data_without_platforms, 'Desired Content Formats', 'Desired Content Formats', split_by=".\n")

    # Analyze and visualize Served Content Platforms
    print("Analyzing: Served Content Platforms Column (From expanded_serp_features_with_platforms.csv)")
    served_content_platforms_table = visualize_and_analyze_columns(data_with_platforms, 'Served Content Platforms', 'Served Content Platforms')

    # Return all processed tables
    return serp_features_table, desired_content_formats_table, served_content_platforms_table

# Call the function and process the data
serp_features_table, desired_content_formats_table, served_content_platforms_table = process_and_visualize_serp_features_fixed_split()

# Save the tables to CSV for inspection if needed
serp_features_table.to_csv('serp_features_table_fixed.csv', index=False)
desired_content_formats_table.to_csv('desired_content_formats_table_fixed.csv', index=False)
served_content_platforms_table.to_csv('served_content_platforms_table_fixed.csv', index=False)
```
